## 1. Clases y Objetos - Los Fundamentos de la Finca 🏡

Una **clase** es como un molde o plano de construcción. En el agro, podemos pensar en ella como:
- Un plano para construir un establo
- Una receta para cultivar un tipo específico de planta
- Un protocolo para criar un tipo de animal

Un **objeto** es una instancia específica de esa clase:
- Tu establo particular con tus vacas específicas
- Tu parcela de maíz con sus características únicas
- Tu tractor con su modelo y año específico

```python
# ========================================
# CULTIVO: NUESTRA PRIMERA CLASE
# ========================================

print("=== CREANDO NUESTRA PRIMERA CLASE: CULTIVO ===")

class Cultivo:
    """
    Clase que representa un cultivo en nuestra finca.
    Cada cultivo tiene nombre, hectáreas, rendimiento esperado y estado.
    """
    
    def __init__(self, nombre, hectareas, rendimiento_por_hectarea):
        # El constructor inicializa los atributos del cultivo
        self.nombre = nombre                           # Tipo de cultivo (ej: "Maíz", "Soja")
        self.hectareas = hectareas                     # Superficie sembrada
        self.rendimiento_por_hectarea = rendimiento_por_hectarea  # Toneladas por hectárea
        self.estado = "Sembrado"                       # Estado inicial del cultivo
        self.dias_desde_siembra = 0                    # Contador de días
    
    def crecer(self, dias):
        """Simula el crecimiento del cultivo durante ciertos días"""
        self.dias_desde_siembra += dias
        if self.dias_desde_siembra < 30:
            self.estado = "Germinando"
        elif self.dias_desde_siembra < 60:
            self.estado = "Creciendo"
        elif self.dias_desde_siembra < 120:
            self.estado = "Floreciendo"
        else:
            self.estado = "Listo para cosecha"
        
        print(f"🌱 {self.nombre}: {dias} días después... Estado: {self.estado}")
    
    def cosechar(self):
        """Realiza la cosecha del cultivo si está listo"""
        if self.estado == "Listo para cosecha":
            produccion_total = self.hectareas * self.rendimiento_por_hectarea
            self.estado = "Cosechado"
            print(f"🚜 ¡Cosecha exitosa de {self.nombre}!")
            print(f"   Producción total: {produccion_total} toneladas")
            return produccion_total
        else:
            print(f"⚠️  {self.nombre} aún no está listo para cosechar. Estado: {self.estado}")
            return 0
    
    def mostrar_info(self):
        """Muestra información completa del cultivo"""
        print(f"📊 Información del cultivo:")
        print(f"   Nombre: {self.nombre}")
        print(f"   Hectáreas: {self.hectareas} ha")
        print(f"   Rendimiento: {self.rendimiento_por_hectarea} ton/ha")
        print(f"   Estado: {self.estado}")
        print(f"   Días desde siembra: {self.dias_desde_siembra}")

# ========================================
# CREANDO OBJETOS CULTIVO
# ========================================

print("\n=== CREANDO CULTIVOS EN NUESTRA FINCA ===")

# Crear diferentes cultivos (objetos de la clase Cultivo)
maiz = Cultivo("Maíz", 50, 8.5)  # 50 hectáreas, 8.5 ton/ha
soja = Cultivo("Soja", 75, 3.2)  # 75 hectáreas, 3.2 ton/ha
trigo = Cultivo("Trigo", 40, 4.1) # 40 hectáreas, 4.1 ton/ha

print("✅ Cultivos creados exitosamente!\n")

# Mostrar información inicial
maiz.mostrar_info()
print()
soja.mostrar_info()
print()
trigo.mostrar_info()
print()
```

```python
# ========================================
# SIMULACIÓN INTERACTIVA DEL CRECIMIENTO
# ========================================

print("=== SIMULANDO EL CICLO DE VIDA DE LOS CULTIVOS ===")

# Simular el paso del tiempo
print("🗓️  Semana 1-2 (0-14 días):")
maiz.crecer(14)
soja.crecer(14)
trigo.crecer(14)

print("\n🗓️  Mes 1-2 (15-45 días):")
maiz.crecer(31)
soja.crecer(31)
trigo.crecer(31)

print("\n🗓️  Mes 2-3 (46-75 días):")
maiz.crecer(30)
soja.crecer(30)
trigo.crecer(30)

print("\n🗓️  Mes 3-4 (76-120 días):")
maiz.crecer(45)
soja.crecer(45)
trigo.crecer(45)

print("\n=== INTENTANDO COSECHAR ===")
produccion_maiz = maiz.cosechar()
produccion_soja = soja.cosechar()
produccion_trigo = trigo.cosechar()

print(f"\n📈 RESUMEN DE PRODUCCIÓN TOTAL:")
print(f"   Maíz: {produccion_maiz} toneladas")
print(f"   Soja: {produccion_soja} toneladas") 
print(f"   Trigo: {produccion_trigo} toneladas")
print(f"   TOTAL: {produccion_maiz + produccion_soja + produccion_trigo} toneladas")
print()
```

## 2. Encapsulación - Protegiendo los Secretos de la Finca 🔒

La **encapsulación** es como tener diferentes niveles de seguridad en tu finca:
- **Público** (`atributo`): Información que todos pueden ver (como el tipo de cultivo)
- **Protegido** (`_atributo`): Solo para la familia y empleados de confianza (como costos internos)
- **Privado** (`__atributo`): Solo para el dueño (como números de cuenta bancaria)

¡Vamos a crear una clase CuentaBancaria para manejar las finanzas de la finca!

```python
# ========================================
# ENCAPSULACIÓN: CUENTA BANCARIA DE LA FINCA
# ========================================

print("=== ENCAPSULACIÓN: PROTEGIENDO DATOS SENSIBLES ===")

class CuentaBancariaFinca:
    """
    Clase que maneja las finanzas de la finca con diferentes niveles de privacidad
    """
    
    def __init__(self, nombre_finca, saldo_inicial):
        # PÚBLICO: Información que todos pueden ver
        self.nombre_finca = nombre_finca
        self.tipo_cuenta = "Cuenta Corriente Agropecuaria"
        
        # PROTEGIDO: Solo para empleados de confianza (prefijo _)
        self._historial_transacciones = []
        self._limite_credito = 50000  # Límite de crédito en dólares
        
        # PRIVADO: Solo para el dueño (prefijo __)
        self.__saldo = saldo_inicial  # Saldo real de la cuenta
        self.__numero_cuenta = "AG-2024-7891"  # Número de cuenta secreto
        self.__pin = "1234"  # PIN de seguridad
    
    def consultar_saldo_publico(self):
        """Información pública: muestra saldo aproximado"""
        # Solo muestra un rango, no el saldo exacto
        if self.__saldo < 10000:
            return "Saldo: Bajo (menos de $10,000)"
        elif self.__saldo < 50000:
            return "Saldo: Medio ($10,000 - $50,000)"
        else:
            return "Saldo: Alto (más de $50,000)"
    
    def consultar_saldo_real(self, pin):
        """Método privado: requiere PIN para ver saldo exacto"""
        if pin == self.__pin:
            return f"💰 Saldo exacto: ${self.__saldo:,.2f}"
        else:
            return "❌ PIN incorrecto. Acceso denegado."
    
    def depositar_venta(self, monto, concepto):
        """Depositar dinero de ventas agrícolas"""
        if monto > 0:
            self.__saldo += monto
            # Agregar al historial protegido
            transaccion = f"DEPÓSITO: +${monto:,.2f} - {concepto}"
            self._historial_transacciones.append(transaccion)
            print(f"✅ Depósito exitoso: ${monto:,.2f}")
            print(f"   Concepto: {concepto}")
        else:
            print("❌ El monto debe ser positivo")
    
    def pagar_gasto(self, monto, concepto):
        """Pagar gastos de la finca"""
        if monto > 0:
            if self.__saldo >= monto:
                self.__saldo -= monto
                transaccion = f"PAGO: -${monto:,.2f} - {concepto}"
                self._historial_transacciones.append(transaccion)
                print(f"✅ Pago realizado: ${monto:,.2f}")
                print(f"   Concepto: {concepto}")
            else:
                print(f"❌ Fondos insuficientes. Saldo disponible: ${self.__saldo:,.2f}")
        else:
            print("❌ El monto debe ser positivo")
    
    def ver_historial_empleado(self):
        """Método protegido: solo empleados de confianza"""
        print("📊 Historial de transacciones (últimas 5):")
        for transaccion in self._historial_transacciones[-5:]:
            print(f"   {transaccion}")
    
    def mostrar_info_publica(self):
        """Información que cualquiera puede ver"""
        print(f"🏦 Información pública de la cuenta:")
        print(f"   Finca: {self.nombre_finca}")
        print(f"   Tipo: {self.tipo_cuenta}")
        print(f"   Estado: {self.consultar_saldo_publico()}")

# ========================================
# DEMOSTRANDO LA ENCAPSULACIÓN
# ========================================

print("\n=== CREANDO CUENTA BANCARIA DE LA FINCA ===")

# Crear cuenta de la finca
finca_los_alamos = CuentaBancariaFinca("Finca Los Álamos", 25000)

# Información pública (cualquiera puede verla)
finca_los_alamos.mostrar_info_publica()
print()

# Intentar acceder al saldo real con PIN incorrecto
print("=== INTENTANDO ACCEDER CON PIN INCORRECTO ===")
print(finca_los_alamos.consultar_saldo_real("0000"))
print()

# Acceder con PIN correcto
print("=== ACCEDIENDO CON PIN CORRECTO ===")
print(finca_los_alamos.consultar_saldo_real("1234"))
print()
```

```python
# ========================================
# SIMULACIÓN DE ACTIVIDADES FINANCIERAS
# ========================================

print("=== SIMULANDO UN MES DE ACTIVIDADES FINANCIERAS ===")

# Ingresos por ventas
print("💰 INGRESOS POR VENTAS:")
finca_los_alamos.depositar_venta(15000, "Venta de maíz - 100 toneladas")
finca_los_alamos.depositar_venta(8500, "Venta de soja - 50 toneladas")
finca_los_alamos.depositar_venta(3200, "Venta de leche - mes completo")

print("\n💸 GASTOS OPERATIVOS:")
finca_los_alamos.pagar_gasto(2500, "Fertilizantes y semillas")
finca_los_alamos.pagar_gasto(1800, "Combustible para maquinaria")
finca_los_alamos.pagar_gasto(3500, "Salarios empleados")
finca_los_alamos.pagar_gasto(800, "Mantenimiento de equipos")

print("\n📊 ESTADO FINAL:")
print(finca_los_alamos.consultar_saldo_real("1234"))

print("\n=== HISTORIAL PARA EMPLEADOS DE CONFIANZA ===")
finca_los_alamos.ver_historial_empleado()

# Demostrar que no podemos acceder directamente a datos privados
print("\n=== DEMOSTRANDO PROTECCIÓN DE DATOS PRIVADOS ===")
try:
    # Esto causará un error porque __saldo es privado
    print(f"Intentando acceder a saldo privado: {finca_los_alamos.__saldo}")
except AttributeError as e:
    print(f"❌ Error esperado: No se puede acceder directamente a datos privados")
    print(f"   Mensaje técnico: {e}")

print("\n✅ La encapsulación está funcionando correctamente!")
print("   Los datos sensibles están protegidos del acceso no autorizado.")
print()
```

## 3. Herencia - La Familia de los Animales de Granja 🐄🐷🐔

La **herencia** es como los rasgos familiares que se transmiten de padres a hijos:
- Una clase **padre** (Animal) tiene características generales
- Las clases **hijas** (Vaca, Cerdo, Pollo) heredan esas características y agregan las suyas

¡Es como cuando una vaca hereda la capacidad de comer y dormir de "Animal", pero agrega su propia capacidad de dar leche!

```python
# ========================================
# HERENCIA: CLASE PADRE "ANIMAL"
# ========================================

print("=== HERENCIA: CREANDO LA FAMILIA DE ANIMALES ===")

class Animal:
    """
    Clase padre que representa un animal genérico de granja.
    Todas las características comunes van aquí.
    """
    
    def __init__(self, nombre, especie, edad, peso):
        # Atributos comunes a todos los animales
        self.nombre = nombre
        self.especie = especie  
        self.edad = edad        # en años
        self.peso = peso        # en kilogramos
        self.salud = "Saludable"
        self.hambre = False
    
    def comer(self, alimento):
        """Método común: todos los animales comen"""
        print(f"🍽️  {self.nombre} está comiendo {alimento}")
        self.hambre = False
    
    def dormir(self):
        """Método común: todos los animales duermen"""
        print(f"😴 {self.nombre} está durmiendo tranquilamente")
    
    def enfermar(self):
        """Método común: todos los animales pueden enfermarse"""
        self.salud = "Enfermo"
        print(f"🤒 {self.nombre} no se siente bien")
    
    def curar(self):
        """Método común: todos los animales pueden curarse"""
        self.salud = "Saludable"
        print(f"💊 {self.nombre} se ha recuperado completamente")
    
    def mostrar_info_basica(self):
        """Información básica del animal"""
        print(f"📋 Información básica:")
        print(f"   Nombre: {self.nombre}")
        print(f"   Especie: {self.especie}")
        print(f"   Edad: {self.edad} años")
        print(f"   Peso: {self.peso} kg")
        print(f"   Salud: {self.salud}")

# ========================================
# HERENCIA: CLASES HIJAS ESPECÍFICAS
# ========================================

class Vaca(Animal):
    """
    Clase hija que hereda de Animal y agrega características específicas de vacas
    """
    
    def __init__(self, nombre, edad, peso, raza):
        # Llamamos al constructor de la clase padre con super()
        super().__init__(nombre, "Vaca", edad, peso)
        # Agregamos atributos específicos de las vacas
        self.raza = raza
        self.litros_leche_diarios = 0
        self.esta_preñada = False
    
    def dar_leche(self):
        """Método específico: solo las vacas dan leche"""
        if self.edad >= 2 and self.salud == "Saludable":
            self.litros_leche_diarios = 25 + (self.peso // 20)  # Basado en peso
            print(f"🥛 {self.nombre} ha dado {self.litros_leche_diarios} litros de leche hoy")
            return self.litros_leche_diarios
        else:
            print(f"❌ {self.nombre} no puede dar leche (muy joven o enferma)")
            return 0
    
    def mugir(self):
        """Método específico: las vacas mugen"""
        print(f"🐄 {self.nombre}: ¡MUUUUU!")

class Cerdo(Animal):
    """
    Clase hija que hereda de Animal y agrega características específicas de cerdos
    """
    
    def __init__(self, nombre, edad, peso):
        super().__init__(nombre, "Cerdo", edad, peso)
        self.peso_ideal_sacrificio = 120  # kg
    
    def revolcarse(self):
        """Método específico: los cerdos se revuelcan en el lodo"""
        print(f"🐷 {self.nombre} se está revolcando felizmente en el lodo")
    
    def gruñir(self):
        """Método específico: los cerdos gruñen"""
        print(f"🐷 {self.nombre}: ¡OINK OINK!")
    
    def verificar_peso_sacrificio(self):
        """Método específico: verificar si está listo para sacrificio"""
        if self.peso >= self.peso_ideal_sacrificio:
            print(f"📊 {self.nombre} está listo para sacrificio ({self.peso} kg)")
            return True
        else:
            print(f"📊 {self.nombre} necesita ganar más peso (actual: {self.peso} kg, ideal: {self.peso_ideal_sacrificio} kg)")
            return False

class Pollo(Animal):
    """
    Clase hija que hereda de Animal y agrega características específicas de pollos
    """
    
    def __init__(self, nombre, edad, peso, sexo):
        super().__init__(nombre, "Pollo", edad, peso)
        self.sexo = sexo  # "Macho" o "Hembra"
        self.huevos_por_semana = 0
    
    def poner_huevo(self):
        """Método específico: solo las gallinas ponen huevos"""
        if self.sexo == "Hembra" and self.edad >= 0.5 and self.salud == "Saludable":
            print(f"🥚 {self.nombre} ha puesto un huevo")
            self.huevos_por_semana += 1
            return 1
        else:
            print(f"❌ {self.nombre} no puede poner huevos")
            return 0
    
    def cacarear(self):
        """Método específico: los pollos cacarean"""
        if self.sexo == "Hembra":
            print(f"🐔 {self.nombre}: ¡COC COC COC!")
        else:
            print(f"🐓 {self.nombre}: ¡KIKIRIKI!")

print("✅ Clases con herencia creadas exitosamente!")
print("   - Animal (clase padre)")
print("   - Vaca, Cerdo, Pollo (clases hijas)")
print()
```

```python
# ========================================
# DEMOSTRANDO LA HERENCIA EN ACCIÓN
# ========================================

print("=== CREANDO NUESTRA GRANJA DE ANIMALES ===")

# Crear diferentes animales
betsy = Vaca("Betsy", 4, 480, "Holstein")
napoleon = Cerdo("Napoleón", 1, 85)  
clementina = Pollo("Clementina", 1, 2.5, "Hembra")
rocco = Pollo("Rocco", 1, 3.1, "Macho")

print("🏡 Animales en la granja:")
print(f"   🐄 {betsy.nombre} - {betsy.especie} {betsy.raza}")
print(f"   🐷 {napoleon.nombre} - {napoleon.especie}")
print(f"   🐔 {clementina.nombre} - {clementina.especie} {clementina.sexo}")
print(f"   🐓 {rocco.nombre} - {rocco.especie} {rocco.sexo}")

print("\n=== USANDO MÉTODOS HEREDADOS (COMUNES A TODOS) ===")

# Todos los animales pueden hacer estas actividades (heredadas de Animal)
betsy.comer("pasto fresco")
napoleon.comer("maíz y desperdicios")
clementina.comer("granos de maíz")
rocco.comer("lombrices")

print()
betsy.dormir()
napoleon.dormir()
clementina.dormir()

print("\n=== USANDO MÉTODOS ESPECÍFICOS DE CADA ANIMAL ===")

# Cada animal tiene sus propios métodos únicos
print("🐄 Actividades específicas de vaca:")
betsy.mugir()
leche_producida = betsy.dar_leche()

print("\n🐷 Actividades específicas de cerdo:")
napoleon.gruñir()
napoleon.revolcarse()
napoleon.verificar_peso_sacrificio()

print("\n🐔 Actividades específicas de pollos:")
clementina.cacarear()
huevos = clementina.poner_huevo()
rocco.cacarear()
rocco.poner_huevo()  # Esto debería fallar porque es macho

print("\n=== INFORMACIÓN DETALLADA DE CADA ANIMAL ===")

print(f"\n📊 {betsy.nombre} (Vaca):")
betsy.mostrar_info_basica()
print(f"   Raza: {betsy.raza}")
print(f"   Leche diaria: {betsy.litros_leche_diarios} litros")

print(f"\n📊 {napoleon.nombre} (Cerdo):")
napoleon.mostrar_info_basica()
print(f"   Peso ideal sacrificio: {napoleon.peso_ideal_sacrificio} kg")

print(f"\n📊 {clementina.nombre} (Pollo):")
clementina.mostrar_info_basica()
print(f"   Sexo: {clementina.sexo}")
print(f"   Huevos esta semana: {clementina.huevos_por_semana}")

print("\n✅ ¡La herencia permite que cada animal sea único pero comparta características básicas!")
print()
```

## 4. Polimorfismo - Un Método, Muchos Comportamientos 🔄

El **polimorfismo** significa "muchas formas". Es como tener un comando "produce()" que:
- Para una vaca significa "dar leche" 🥛
- Para un pollo significa "poner huevos" 🥚  
- Para un cerdo significa "ganar peso para carne" 🥩
- Para un cultivo significa "crecer y dar frutos" 🌽

¡El mismo comando, diferentes resultados según quien lo ejecute!

```python
# ========================================
# POLIMORFISMO: AGREGANDO MÉTODO "PRODUCIR"
# ========================================

print("=== POLIMORFISMO: EL MISMO MÉTODO, DIFERENTES COMPORTAMIENTOS ===")

# Modificamos nuestras clases para agregar el método "producir"
# Cada clase lo implementará de manera diferente

class AnimalProductivo(Animal):
    """Clase base con método producir que será sobrescrito"""
    
    def producir(self):
        """Método genérico que cada animal implementará diferente"""
        print(f"{self.nombre} está produciendo algo...")
        return 0

class VacaProductiva(AnimalProductivo):
    """Vaca que implementa su propia versión de producir"""
    
    def __init__(self, nombre, edad, peso, raza):
        super().__init__(nombre, "Vaca", edad, peso)
        self.raza = raza
        self.litros_leche_diarios = 0
    
    def producir(self):
        """Para las vacas, producir = dar leche"""
        if self.edad >= 2 and self.salud == "Saludable":
            self.litros_leche_diarios = 25 + (self.peso // 20)
            print(f"🥛 {self.nombre} produce {self.litros_leche_diarios} litros de leche")
            return self.litros_leche_diarios
        else:
            print(f"❌ {self.nombre} no puede producir leche")
            return 0

class CerdoProductivo(AnimalProductivo):
    """Cerdo que implementa su propia versión de producir"""
    
    def __init__(self, nombre, edad, peso):
        super().__init__(nombre, "Cerdo", edad, peso)
        self.ganancia_peso_semanal = 5
    
    def producir(self):
        """Para los cerdos, producir = ganar peso para carne"""
        self.peso += self.ganancia_peso_semanal
        kilos_carne = self.peso * 0.75  # 75% del peso vivo es carne aprovechable
        print(f"🥩 {self.nombre} ha ganado peso - Carne potencial: {kilos_carne:.1f} kg")
        return kilos_carne

class PolloProductivo(AnimalProductivo):
    """Pollo que implementa su propia versión de producir"""
    
    def __init__(self, nombre, edad, peso, sexo):
        super().__init__(nombre, "Pollo", edad, peso)
        self.sexo = sexo
        self.huevos_semanales = 0
    
    def producir(self):
        """Para los pollos, producir = poner huevos (solo hembras)"""
        if self.sexo == "Hembra" and self.edad >= 0.5 and self.salud == "Saludable":
            huevos = 6  # 6 huevos por semana
            self.huevos_semanales += huevos
            print(f"🥚 {self.nombre} produce {huevos} huevos esta semana")
            return huevos
        else:
            print(f"❌ {self.nombre} no produce huevos (macho o muy joven)")
            return 0

class CultivoProductivo:
    """Cultivo que también puede producir - diferentes tipos de clases en una lista"""
    
    def __init__(self, nombre, hectareas, rendimiento):
        self.nombre = nombre
        self.hectareas = hectareas
        self.rendimiento_por_hectarea = rendimiento
        self.estado = "Maduro"
    
    def producir(self):
        """Para los cultivos, producir = dar frutos/granos"""
        if self.estado == "Maduro":
            produccion = self.hectareas * self.rendimiento_por_hectarea
            print(f"🌽 {self.nombre} produce {produccion} toneladas de grano")
            return produccion
        else:
            print(f"❌ {self.nombre} no está listo para producir")
            return 0

# ========================================
# DEMOSTRANDO EL POLIMORFISMO EN ACCIÓN
# ========================================

print("\n=== CREANDO ELEMENTOS PRODUCTIVOS DE LA FINCA ===")

# Crear diferentes elementos productivos
lucia = VacaProductiva("Lucía", 3, 520, "Jersey")
porky = CerdoProductivo("Porky", 1, 90)
clara = PolloProductivo("Clara", 1, 2.8, "Hembra")
pepe = PolloProductivo("Pepe", 1, 3.2, "Macho")
maiz_campo_sur = CultivoProductivo("Maíz Campo Sur", 30, 9.5)

# Lista con todos los elementos productivos (¡diferentes tipos de objetos!)
elementos_productivos = [lucia, porky, clara, pepe, maiz_campo_sur]

print("🏭 Elementos productivos en la finca:")
for elemento in elementos_productivos:
    print(f"   📦 {elemento.nombre}")

print("\n=== ¡POLIMORFISMO EN ACCIÓN! ===")
print("Llamando al método 'producir()' en cada elemento...")
print("(El mismo método, diferentes comportamientos)")

produccion_total = {}
for elemento in elementos_productivos:
    print(f"\n🔄 Producción de {elemento.nombre}:")
    resultado = elemento.producir()
    
    # Clasificar la producción por tipo
    if isinstance(elemento, VacaProductiva):
        produccion_total['leche'] = produccion_total.get('leche', 0) + resultado
    elif isinstance(elemento, CerdoProductivo):
        produccion_total['carne'] = produccion_total.get('carne', 0) + resultado
    elif isinstance(elemento, PolloProductivo):
        produccion_total['huevos'] = produccion_total.get('huevos', 0) + resultado
    elif isinstance(elemento, CultivoProductivo):
        produccion_total['granos'] = produccion_total.get('granos', 0) + resultado

print("\n📊 RESUMEN DE PRODUCCIÓN SEMANAL:")
print("=" * 40)
for tipo, cantidad in produccion_total.items():
    if tipo == 'leche':
        print(f"🥛 Leche: {cantidad} litros")
    elif tipo == 'carne':
        print(f"🥩 Carne potencial: {cantidad:.1f} kg")
    elif tipo == 'huevos':
        print(f"🥚 Huevos: {cantidad} unidades")
    elif tipo == 'granos':
        print(f"🌽 Granos: {cantidad} toneladas")

print("\n✅ ¡POLIMORFISMO EXITOSO!")
print("   Un solo método 'producir()' funcionó diferente para cada tipo de elemento")
print("   ¡Esto es la magia del polimorfismo! 🎭")
print()
```

## 5. Proyecto Final: Sistema de Gestión de Finca Completo 🏡

¡Ahora vamos a integrar TODOS los conceptos en un sistema real de gestión de finca! 

Crearemos una clase `Finca` que maneje:
- **Cultivos** y **Animales** (Objetos y Clases)
- **Finanzas privadas** (Encapsulación) 
- **Diferentes tipos de elementos** (Herencia)
- **Producción unificada** (Polimorfismo)

¡Un sistema completo y funcional! 🚀

```python
# ========================================
# SISTEMA COMPLETO DE GESTIÓN DE FINCA
# ========================================

print("=== SISTEMA INTEGRAL DE GESTIÓN DE FINCA ===")

class SistemaFinca:
    """
    Sistema completo que integra TODOS los conceptos de POO:
    - Clases y Objetos: Maneja múltiples elementos
    - Encapsulación: Protege datos financieros sensibles
    - Herencia: Usa jerarquías de cultivos y animales
    - Polimorfismo: Método único para toda la producción
    """
    
    def __init__(self, nombre_finca, propietario, presupuesto_inicial):
        # INFORMACIÓN PÚBLICA
        self.nombre_finca = nombre_finca
        self.propietario = propietario
        self.fecha_creacion = "2024"
        
        # DATOS PROTEGIDOS (para empleados de confianza)
        self._elementos_productivos = []  # Lista de cultivos y animales
        self._historial_produccion = []
        
        # DATOS PRIVADOS (solo propietario)
        self.__presupuesto = presupuesto_inicial
        self.__gastos_totales = 0
        self.__ingresos_totales = 0
        self.__pin_acceso = "AGRO2024"
    
    # ========================================
    # MÉTODOS PÚBLICOS (CUALQUIERA PUEDE USAR)
    # ========================================
    
    def agregar_elemento_productivo(self, elemento):
        """Agregar un cultivo o animal a la finca"""
        self._elementos_productivos.append(elemento)
        print(f"✅ {elemento.nombre} agregado a la finca {self.nombre_finca}")
    
    def mostrar_inventario(self):
        """Mostrar todos los elementos de la finca"""
        print(f"📦 INVENTARIO DE {self.nombre_finca.upper()}:")
        print(f"   Propietario: {self.propietario}")
        print(f"   Total de elementos: {len(self._elementos_productivos)}")
        
        for i, elemento in enumerate(self._elementos_productivos, 1):
            tipo = type(elemento).__name__
            print(f"   {i}. {elemento.nombre} ({tipo})")
    
    def ejecutar_produccion_completa(self):
        """POLIMORFISMO: Un método que funciona para todos los elementos"""
        print(f"🏭 EJECUTANDO PRODUCCIÓN EN {self.nombre_finca.upper()}")
        print("=" * 50)
        
        produccion_total = {
            'leche': 0,
            'huevos': 0, 
            'carne': 0,
            'granos': 0
        }
        
        for elemento in self._elementos_productivos:
            print(f"\n🔄 Produciendo en: {elemento.nombre}")
            resultado = elemento.producir()
            
            # Clasificar producción y sumar al total
            if isinstance(elemento, VacaProductiva):
                produccion_total['leche'] += resultado
            elif isinstance(elemento, PolloProductivo) and elemento.sexo == "Hembra":
                produccion_total['huevos'] += resultado
            elif isinstance(elemento, CerdoProductivo):
                produccion_total['carne'] += resultado
            elif isinstance(elemento, CultivoProductivo):
                produccion_total['granos'] += resultado
        
        # Guardar en historial protegido
        registro_produccion = f"Producción del día: {produccion_total}"
        self._historial_produccion.append(registro_produccion)
        
        return produccion_total
    
    # ========================================
    # MÉTODOS PROTEGIDOS (SOLO EMPLEADOS DE CONFIANZA)
    # ========================================
    
    def _ver_historial_produccion(self):
        """Método protegido: historial de producción"""
        print("📊 HISTORIAL DE PRODUCCIÓN (últimos 5 registros):")
        for registro in self._historial_produccion[-5:]:
            print(f"   📈 {registro}")
    
    def _calcular_ingresos_estimados(self, produccion):
        """Método protegido: calcular ingresos basados en precios de mercado"""
        precios = {
            'leche': 0.45,    # dólares por litro
            'huevos': 0.25,   # dólares por huevo
            'carne': 3.50,    # dólares por kg
            'granos': 200     # dólares por tonelada
        }
        
        ingresos_estimados = 0
        print("💰 INGRESOS ESTIMADOS:")
        
        for producto, cantidad in produccion.items():
            if cantidad > 0:
                ingreso = cantidad * precios[producto]
                ingresos_estimados += ingreso
                print(f"   {producto.capitalize()}: {cantidad} unidades × ${precios[producto]} = ${ingreso:.2f}")
        
        print(f"   💵 TOTAL ESTIMADO: ${ingresos_estimados:.2f}")
        return ingresos_estimados
    
    # ========================================
    # MÉTODOS PRIVADOS (SOLO PROPIETARIO)
    # ========================================
    
    def __verificar_pin(self, pin):
        """Verificación de seguridad privada"""
        return pin == self.__pin_acceso
    
    def consultar_finanzas_completas(self, pin):
        """Método privado: información financiera completa"""
        if not self.__verificar_pin(pin):
            print("❌ PIN INCORRECTO - Acceso denegado a información financiera")
            return None
        
        print("🏦 ESTADO FINANCIERO COMPLETO:")
        print(f"   💰 Presupuesto disponible: ${self.__presupuesto:.2f}")
        print(f"   📈 Ingresos totales: ${self.__ingresos_totales:.2f}")
        print(f"   📉 Gastos totales: ${self.__gastos_totales:.2f}")
        print(f"   💵 Balance neto: ${self.__presupuesto + self.__ingresos_totales - self.__gastos_totales:.2f}")
    
    def registrar_venta(self, pin, monto, concepto):
        """Método privado: registrar venta de producción"""
        if not self.__verificar_pin(pin):
            print("❌ PIN INCORRECTO - No se puede registrar venta")
            return False
        
        self.__ingresos_totales += monto
        self.__presupuesto += monto
        print(f"✅ Venta registrada: ${monto:.2f} - {concepto}")
        return True
    
    def pagar_gasto(self, pin, monto, concepto):
        """Método privado: pagar gastos de la finca"""
        if not self.__verificar_pin(pin):
            print("❌ PIN INCORRECTO - No se puede procesar pago")
            return False
        
        if self.__presupuesto >= monto:
            self.__presupuesto -= monto
            self.__gastos_totales += monto
            print(f"✅ Pago realizado: ${monto:.2f} - {concepto}")
            return True
        else:
            print(f"❌ Fondos insuficientes. Disponible: ${self.__presupuesto:.2f}")
            return False

print("✅ Sistema completo de gestión de finca creado!")
print("   Integra los 4 pilares de POO en un sistema real")
print()
```

```python
# ========================================
# SIMULACIÓN COMPLETA DEL SISTEMA
# ========================================

print("=== SIMULACIÓN COMPLETA: FINCA 'LOS ÁLAMOS' ===")

# Crear el sistema de finca
finca_los_alamos = SistemaFinca("Finca Los Álamos", "Juan Pérez", 50000)

print(f"🏡 {finca_los_alamos.nombre_finca} creada exitosamente")
print(f"   Propietario: {finca_los_alamos.propietario}")

# Agregar elementos productivos usando HERENCIA
print("\n=== POBLANDO LA FINCA ===")

# Crear y agregar animales
vaca_marta = VacaProductiva("Marta", 4, 500, "Holstein")
vaca_sofia = VacaProductiva("Sofía", 3, 480, "Jersey")
cerdo_benito = CerdoProductivo("Benito", 1, 95)
cerdo_rosa = CerdoProductivo("Rosa", 1, 88)
gallina_carmen = PolloProductivo("Carmen", 1, 2.5, "Hembra")
gallina_lucia = PolloProductivo("Lucía", 1, 2.8, "Hembra")
gallo_rodrigo = PolloProductivo("Rodrigo", 1, 3.5, "Macho")

# Crear y agregar cultivos
maiz_norte = CultivoProductivo("Maíz Campo Norte", 25, 9.2)
soja_sur = CultivoProductivo("Soja Campo Sur", 40, 3.5)
trigo_este = CultivoProductivo("Trigo Campo Este", 20, 4.8)

# Agregar todos al sistema
elementos = [vaca_marta, vaca_sofia, cerdo_benito, cerdo_rosa, 
             gallina_carmen, gallina_lucia, gallo_rodrigo,
             maiz_norte, soja_sur, trigo_este]

for elemento in elementos:
    finca_los_alamos.agregar_elemento_productivo(elemento)

print("\n=== INVENTARIO ACTUAL ===")
finca_los_alamos.mostrar_inventario()

print("\n=== EJECUTANDO PRODUCCIÓN SEMANAL ===")
# POLIMORFISMO en acción: un método para todos los elementos
produccion_semanal = finca_los_alamos.ejecutar_produccion_completa()

print(f"\n📊 RESUMEN DE PRODUCCIÓN SEMANAL:")
print("=" * 45)
for producto, cantidad in produccion_semanal.items():
    if cantidad > 0:
        if producto == 'leche':
            print(f"🥛 Leche: {cantidad} litros")
        elif producto == 'huevos':
            print(f"🥚 Huevos: {cantidad} unidades")
        elif producto == 'carne':
            print(f"🥩 Carne potencial: {cantidad:.1f} kg")
        elif producto == 'granos':
            print(f"🌽 Granos: {cantidad} toneladas")

print("\n=== ANÁLISIS FINANCIERO (EMPLEADO DE CONFIANZA) ===")
# Métodos protegidos
ingresos_estimados = finca_los_alamos._calcular_ingresos_estimados(produccion_semanal)
finca_los_alamos._ver_historial_produccion()

print("\n=== OPERACIONES FINANCIERAS (SOLO PROPIETARIO) ===")
# Registrar ventas con PIN correcto
finca_los_alamos.registrar_venta("AGRO2024", ingresos_estimados, "Venta producción semanal")

# Pagar algunos gastos
finca_los_alamos.pagar_gasto("AGRO2024", 1500, "Alimento para animales")
finca_los_alamos.pagar_gasto("AGRO2024", 800, "Combustible para maquinaria")
finca_los_alamos.pagar_gasto("AGRO2024", 1200, "Fertilizantes")

# Ver estado financiero completo
print("\n📊 ESTADO FINANCIERO COMPLETO:")
finca_los_alamos.consultar_finanzas_completas("AGRO2024")

print("\n=== INTENTANDO ACCESO NO AUTORIZADO ===")
# Demostrar seguridad de encapsulación
finca_los_alamos.consultar_finanzas_completas("1234")  # PIN incorrecto

print("\n🎉 ¡SIMULACIÓN COMPLETA EXITOSA!")
print("=" * 50)
print("✅ CONCEPTOS DE POO DEMOSTRADOS:")
print("   🏗️  CLASES Y OBJETOS: Múltiples elementos con sus propiedades")
print("   🔒 ENCAPSULACIÓN: Datos protegidos con diferentes niveles de acceso")
print("   👨‍👩‍👧‍👦 HERENCIA: Animales y cultivos heredan de clases base")
print("   🔄 POLIMORFISMO: Método 'producir()' funciona diferente para cada tipo")
print("\n🚜 ¡Has completado tu curso de POO aplicado al agro!")
print("   Ahora puedes crear sistemas complejos usando estos principios")
print()
```

- **Qué aprendiste**: Clases hijas heredan de clases padres
- **Ejemplo agrícola**: `Animal` → `Vaca`, `Cerdo`, `Pollo` con características compartidas

#### 4️⃣ **Polimorfismo** 
- **Qué aprendiste**: Un método, múltiples comportamientos
- **Ejemplo agrícola**: `producir()` da leche, huevos, carne o granos según el elemento
