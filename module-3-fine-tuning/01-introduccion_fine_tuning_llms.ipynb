# 🌱 Parte 1: Conceptos Fundamentales (30-40 min)

## ¿Qué es el Fine-Tuning?

Imaginemos que tenés un empleado muy inteligente que sabe de todo un poco, pero necesitás que se especialice en agricultura argentina. Tenés tres opciones principales:

### 1. **Prompt Engineering**
Le das instrucciones muy detalladas cada vez que lo consultás:
```
"Sos un experto en agricultura argentina, considerá el clima de la Pampa Húmeda..."
```
**Ventajas:** Es rápido y fácil de implementar
**Desventajas:** Tenés que repetir las instrucciones constantemente

### 2. **RAG (Retrieval-Augmented Generation)**
Le das acceso a una biblioteca especializada con información actualizada:
```
Modelo + Base de datos con info agrícola → Respuesta informada
```
**Ventajas:** Información actualizada y específica del dominio
**Desventajas:** Dependés de tener buena información en la base de datos

### 3. **Fine-Tuning**
Le das entrenamiento específico para que internalice el conocimiento agrícola:
```
Modelo base + Datos de entrenamiento agrícolas → Modelo especializado
```
**Ventajas:** El modelo absorbe el conocimiento específico de manera permanente
**Desventajas:** Requiere más tiempo, datos y recursos computacionales

## 🤔 ¿Cuándo Usar Cada Técnica?

### Usá **Prompt Engineering** cuando:
- Necesitás resultados rápidos y no tenés mucho tiempo para invertir
- Tenés tareas simples o consultas puntuales
- No querés invertir en recursos computacionales adicionales

**Ejemplo agrícola:** Análisis ocasional de datos climáticos para una campaña específica

### Usá **RAG** cuando:
- Necesitás información que se actualiza constantemente
- Ya tenés una buena base de conocimiento estructurada
- Querés mantener la información separada del modelo para fácil actualización

**Ejemplo agrícola:** Sistema de consultas sobre precios de commodities que cambian diariamente

### Usá **Fine-Tuning** cuando:
- Necesitás un comportamiento muy específico y consistente
- Querés que el modelo razone como un experto del dominio
- Tenés suficientes datos de calidad para entrenar (mínimo 500-1000 ejemplos)
- Vas a usar el modelo repetidamente para tareas similares

**Ejemplo agrícola:** Asistente que siempre responde como un ingeniero agrónomo argentino especializado

## 🌾 Casos de Uso en Agricultura

### Ejemplos donde Fine-Tuning es Ideal:

**1. Diagnóstico de Enfermedades**
- Entrenar el modelo con miles de descripciones de síntomas y sus diagnósticos correspondientes
- El modelo aprende a reconocer patrones específicos de cada enfermedad
- Resultado: Diagnósticos más precisos y consistentes, similar al razonamiento de un fitopatólogo

**2. Recomendaciones de Manejo**
- Entrenar con decisiones históricas de productores exitosos y sus justificaciones técnicas
- El modelo internaliza las mejores prácticas regionales y su aplicación contextual
- Resultado: Consejos adaptados al contexto local y condiciones específicas

**3. Interpretación de Análisis de Suelos**
- Entrenar con miles de análisis de laboratorio y sus interpretaciones agronómicas
- El modelo aprende a interpretar los valores numéricos como lo haría un especialista
- Resultado: Explicaciones claras y recomendaciones precisas para productores

**4. Comunicación con Productores**
- Entrenar con conversaciones reales entre técnicos y productores
- El modelo internaliza el "lenguaje del campo" y la forma apropiada de comunicar conceptos técnicos
- Resultado: Comunicación más efectiva, natural y culturalmente apropiada

# ⚙️ Parte 2: El Proceso Paso a Paso (20-30 min)

## 🗺️ Roadmap del Fine-Tuning

El fine-tuning sigue un proceso estructurado:

```
📊 Datos → 🧹 Preparación → 🎯 Entrenamiento → 📈 Evaluación → 🚀 Despliegue
```

## 1️⃣ Preparación de Datos

### ¿Qué Datos Necesitás?

Para fine-tuning necesitás **pares de ejemplos** bien estructurados:
- **Entrada:** La pregunta o contexto que va a recibir el modelo
- **Salida:** La respuesta ideal que esperás que genere

### Ejemplo Agrícola:

**Dato Malo (muy general):**
```
"El maíz tiene manchas amarillas"
```

**Dato Bueno (específico y completo):**
```json
{
  "entrada": "Productor de Pergamino reporta manchas amarillas en hojas de maíz V6, principalmente en hojas inferiores, con clima húmedo los últimos 15 días",
  "salida": "Posible roya común (Puccinia sorghi). Recomiendo: 1) Confirmar diagnóstico buscando pústulas naranjas, 2) Monitorear propagación, 3) Considerar fungicida si supera umbral del 5% de área foliar afectada. En V6 aún hay tiempo para tratamiento efectivo."
}
```

### Principios de Calidad vs Cantidad:
- **100 ejemplos excelentes** son más valiosos que **1000 ejemplos mediocres**
- Es fundamental tener datos consistentes y bien etiquetados
- Debés cubrir la diversidad de casos que el modelo va a enfrentar en producción

## 2️⃣ Técnicas de Fine-Tuning

### Full Fine-Tuning vs LoRA

#### **Full Fine-Tuning**
Modificás todos los parámetros del modelo durante el entrenamiento
- **Ventajas:** Máxima capacidad de personalización y adaptación
- **Desventajas:** Requiere mucha memoria GPU y tiempo de entrenamiento
- **Analogía:** Sería como reeducar completamente a un empleado desde cero

#### **LoRA (Low-Rank Adaptation)**
Solo modificás pequeñas "adaptaciones" que se agregan al modelo original
- **Ventajas:** Rápido, eficiente en memoria, mantiene conocimiento base
- **Desventajas:** Menor flexibilidad de personalización profunda
- **Analogía:** Es como darle una especialización específica a un empleado ya capacitado

### ¿Cuál conviene elegir?
- **LoRA:** Recomendado para la mayoría de casos, especialmente cuando empezás
- **Full Fine-Tuning:** Solo consideralo si LoRA no logra los resultados que necesitás

## 3️⃣ Proceso de Entrenamiento

### Hiperparámetros Clave (Simplificado):

**Learning Rate (Tasa de Aprendizaje)**
- Determina qué tan rápido aprende el modelo de los nuevos datos
- Valor muy alto: aprende rápido pero puede "olvidar" conocimiento previo (catastrofic forgetting)
- Valor muy bajo: aprende lento pero de manera más estable
- **Recomendación:** Empezar con los valores sugeridos por la herramienta (típicamente 2e-4 para LoRA)

**Batch Size**
- Define cuántos ejemplos procesa el modelo simultáneamente
- Batch más grande: entrenamiento más estable pero requiere más memoria
- Batch más pequeño: menor uso de memoria pero mayor variabilidad
- **Recomendación:** Usar el batch size máximo que permita tu GPU sin quedarse sin memoria

**Epochs**
- Número de veces que el modelo pasa por todo el dataset de entrenamiento
- Muy pocos epochs: el modelo no aprende suficientemente
- Demasiados epochs: el modelo "memoriza" en lugar de generalizar (overfitting)
- **Recomendación:** Monitorear métricas de validación y parar cuando dejen de mejorar

## 4️⃣ Evaluación del Modelo

### ¿Cómo Saber si Funciona?

**Evaluación Cuantitativa**
- Métricas automáticas como BLEU, ROUGE, perplexity, etc.
- Útil para comparar diferentes versiones del modelo de manera objetiva
- Limitación: no siempre refleja la calidad real percibida por usuarios

**Evaluación Cualitativa**
- Revisión manual de respuestas por expertos del dominio
- Evaluación de casos extremos o difíciles (edge cases)
- Más relevante para determinar la utilidad práctica del modelo

### Test Set Específico para Agricultura:
```
Casos de prueba que deberías incluir:
• Diagnóstico de enfermedades comunes
• Recomendaciones de fertilización
• Interpretación de análisis de suelos
• Manejo de plagas regionales
• Planificación de siembra según clima
```

## 📝 Ejemplo Práctico: Asistente Agrícola

Vamos a crear un modelo que responda consultas agrícolas como un ingeniero agrónomo argentino.

### Paso 1: Importar librerías y cargar modelo base

````python
from unsloth import FastLanguageModel
import torch

# Configuración de parámetros principales
max_seq_length = 2048  # Longitud máxima de secuencia en tokens (aprox 1500-2000 palabras)
dtype = None  # Permite auto-detección del mejor tipo de datos para tu GPU
load_in_4bit = True  # Activa quantización 4-bit para ahorrar ~75% de memoria

# Cargar modelo base preentrenado
# Llama 3.1 8B con quantización 4-bit: buen balance entre calidad y eficiencia
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/llama-3.1-8b-bnb-4bit",  # Modelo base de Meta AI optimizado
    max_seq_length=max_seq_length,  # Longitud máxima de contexto
    dtype=dtype,  # Tipo de datos (auto-detectado)
    load_in_4bit=load_in_4bit,  # Quantización para eficiencia de memoria
)

print("Modelo cargado exitosamente")
print(f"Modelo: {model.config.name_or_path}")
print(f"Parámetros: ~8B (quantizados a 4-bit)")
````

### Paso 2: Configurar para Fine-Tuning

````python
# Configurar el modelo para LoRA (Low-Rank Adaptation)
model = FastLanguageModel.get_peft_model(
    model,
    r=16,  # Rank de LoRA: controla complejidad vs eficiencia (típico: 8-64)
    
    # Módulos objetivo: capas del transformer que se van a adaptar
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",  # Attention layers
                   "gate_proj", "up_proj", "down_proj"],  # Feed-forward layers
    
    lora_alpha=16,  # Factor de escalado para las adaptaciones LoRA (típico: igual a 'r')
    lora_dropout=0,  # Dropout en capas LoRA (0 = sin dropout para estabilidad)
    bias="none",  # No entrenar bias ("none" | "all" | "lora_only")
    
    # Configuraciones avanzadas de Unsloth
    use_gradient_checkpointing="unsloth",  # Optimización de memoria específica de Unsloth
    random_state=3407,  # Semilla para reproducibilidad
    use_rslora=False,  # Rank-Stabilized LoRA (experimental)
    loftq_config=None,  # LoftQ quantization (None = usar quantización estándar)
)

print("Configuración LoRA aplicada")
print(f"Parámetros entrenables: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
print(f"Parámetros totales: {sum(p.numel() for p in model.parameters()):,}")
````

### Paso 3: Preparar Datos de Entrenamiento

Para este ejemplo, vamos a crear un pequeño dataset de consultas agrícolas:

````python
# Template de formato Alpaca para estructurar conversaciones
# Este formato es estándar para fine-tuning de modelos conversacionales
alpaca_prompt = """A continuación hay una instrucción que describe una tarea, junto con una entrada que proporciona más contexto. Escribí una respuesta que complete adecuadamente la solicitud.

### Instrucción:
{}

### Entrada:
{}

### Respuesta:
{}"""

# Dataset de ejemplo con consultas agrícolas argentinas
# En un proyecto real, tendrías 500-5000 ejemplos similares
ejemplos_agricolas = [
    {
        "instruccion": "Actúa como un ingeniero agrónomo especializado en cultivos de la región pampeana argentina y proporciona recomendaciones técnicas.",
        "entrada": "Mi campo de soja en Pergamino presenta hojas amarillentas en las plantas, especialmente en los sectores bajos del lote. ¿Qué puede estar pasando?",
        "salida": "Las hojas amarillentas en soja, especialmente en sectores bajos, pueden indicar varias problemáticas: 1) Encharcamiento: Los sectores bajos retienen más agua, causando asfixia radicular. Solución: mejorar drenaje. 2) Deficiencia de nitrógeno: La bacteria fijadora puede estar comprometida por exceso de humedad. 3) Enfermedades fúngicas como Phytophthora o Rhizoctonia, favorecidas por humedad. Recomiendo: inspección radicular, análisis de suelo en zonas afectadas y aplicación de fungicida preventivo si es necesario."
    },
    {
        "instruccion": "Actúa como un ingeniero agrónomo especializado en cultivos de la región pampeana argentina y proporciona recomendaciones técnicas.",
        "entrada": "¿Cuál es el momento óptimo para sembrar maíz en la zona de Marcos Juárez, Córdoba?",
        "salida": "En Marcos Juárez, el momento óptimo para siembra de maíz depende del objetivo: 1) Maíz temprano: Desde 15 de septiembre hasta 15 de octubre, aprovecha mejor las lluvias y evita stress hídrico en floración. 2) Maíz tardío: Desde noviembre hasta diciembre, menor riesgo de heladas pero mayor riesgo hídrico. Para la zona centro de Córdoba recomiendo siembra temprana (octubre) con híbridos de 115-125 días, densidad 70.000-75.000 plantas/ha. Considerar fecha de heladas (promedio 15 de mayo) para planificar cosecha."
    }
]

# Función para formatear cada ejemplo según el template Alpaca
def formatear_ejemplo(ejemplo):
    return alpaca_prompt.format(
        ejemplo["instruccion"],  # Rol y contexto del modelo
        ejemplo["entrada"],     # Pregunta o consulta del usuario
        ejemplo["salida"]       # Respuesta esperada del modelo
    ) + tokenizer.eos_token  # Token de fin de secuencia

# Aplicar formato a todos los ejemplos
# En un caso real, cargarías desde archivo CSV/JSON con pandas
dataset_formateado = [formatear_ejemplo(ej) for ej in ejemplos_agricolas]

print(f"Preparados {len(dataset_formateado)} ejemplos de entrenamiento")
print(f"Longitud promedio: {sum(len(tokenizer.encode(text)) for text in dataset_formateado) // len(dataset_formateado)} tokens")
print("\nEjemplo de formato:")
print(dataset_formateado[0][:400] + "...")
````

### Paso 4: Configurar Entrenamiento

````python
from trl import SFTTrainer  # Supervised Fine-Tuning Trainer
from transformers import TrainingArguments
from datasets import Dataset

# Convertir lista de strings a formato Dataset de HuggingFace
dataset = Dataset.from_dict({"text": dataset_formateado})

# Configurar el entrenador con parámetros optimizados
trainer = SFTTrainer(
    model=model,  # Modelo con LoRA configurado
    tokenizer=tokenizer,  # Tokenizador del modelo base
    train_dataset=dataset,  # Dataset de entrenamiento
    dataset_text_field="text",  # Campo que contiene el texto a entrenar
    max_seq_length=max_seq_length,  # Longitud máxima de secuencia
    dataset_num_proc=2,  # Procesos paralelos para procesar dataset
    packing=False,  # No empaquetar múltiples ejemplos en una secuencia
    
    # Argumentos de entrenamiento
    args=TrainingArguments(
        per_device_train_batch_size=2,  # Ejemplos por batch por GPU (ajustar según memoria)
        gradient_accumulation_steps=4,  # Acumular gradientes (batch efectivo = 2*4 = 8)
        warmup_steps=5,  # Pasos de calentamiento para learning rate
        max_steps=10,  # Total de pasos de entrenamiento (demo: 10, real: 60-200)
        
        learning_rate=2e-4,  # Tasa de aprendizaje (típica para LoRA: 1e-4 a 5e-4)
        
        # Configuración de precisión numérica
        fp16=not torch.cuda.is_bf16_supported(),  # Usar FP16 si BF16 no está disponible
        bf16=torch.cuda.is_bf16_supported(),      # Usar BF16 si está disponible (mejor)
        
        # Logging y guardado
        logging_steps=1,  # Frecuencia de logging
        output_dir="outputs",  # Directorio de salida
        
        # Optimizador y regularización
        optim="adamw_8bit",  # Optimizador AdamW con 8-bit (ahorra memoria)
        weight_decay=0.01,   # Regularización L2
        lr_scheduler_type="linear",  # Tipo de scheduler de learning rate
        
        seed=3407,  # Semilla para reproducibilidad
    ),
)

print("Trainer configurado")
print(f"Batch size efectivo: {2 * 4} (per_device_batch_size * gradient_accumulation_steps)")
print(f"Pasos totales: 10 (solo para demostración)")
````

### Paso 5: Entrenar el Modelo

````python
# Monitoreo de recursos antes del entrenamiento
gpu_stats = torch.cuda.get_device_properties(0)  # Obtener información de la GPU
start_gpu_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)  # Memoria en GB
max_memory = round(gpu_stats.total_memory / 1024 / 1024 / 1024, 3)  # Memoria total GPU

print(f"GPU: {gpu_stats.name}")
print(f"Memoria máxima disponible: {max_memory} GB")
print(f"Memoria usada antes del entrenamiento: {start_gpu_memory} GB")
print(f"Memoria libre: {max_memory - start_gpu_memory:.1f} GB")

# Iniciar proceso de entrenamiento
print("\nIniciando entrenamiento...")
print("Esto puede tomar varios minutos dependiendo de tu GPU")

# Entrenar el modelo (aquí es donde ocurre la magia del fine-tuning)
trainer_stats = trainer.train()

# Estadísticas finales de memoria y rendimiento
used_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
used_memory_for_lora = round(used_memory - start_gpu_memory, 3)
used_percentage = round(used_memory / max_memory * 100, 3)
lora_percentage = round(used_memory_for_lora / max_memory * 100, 3)

print(f"\nEntrenamiento completado exitosamente!")
print(f"Memoria total usada: {used_memory} GB ({used_percentage}% de la GPU)")
print(f"Memoria adicional para LoRA: {used_memory_for_lora} GB ({lora_percentage}%)")
print(f"Tiempo total de entrenamiento: {trainer_stats.metrics['train_runtime']:.2f} segundos")
print(f"Loss final: {trainer_stats.metrics['train_loss']:.4f}")
````

### Paso 6: Probar el Modelo Entrenado

````python
# Preparar el modelo para inferencia (desactivar entrenamiento)
FastLanguageModel.for_inference(model)

# Función para hacer consultas al modelo entrenado
def consultar_modelo(pregunta, max_tokens=256, temperatura=0.7):
    """
    Función para consultar el modelo fine-tuneado
    
    Args:
        pregunta (str): Consulta agrícola del usuario
        max_tokens (int): Máximo número de tokens a generar
        temperatura (float): Creatividad (0.1=conservador, 1.0=creativo)
    """
    # Formatear la consulta usando el mismo template de entrenamiento
    prompt = alpaca_prompt.format(
        "Actúa como un ingeniero agrónomo especializado en cultivos de la región pampeana argentina y proporciona recomendaciones técnicas.",
        pregunta,
        ""  # Respuesta vacía para que el modelo la complete
    )
    
    # Tokenizar la entrada
    inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
    
    # Generar respuesta
    outputs = model.generate(
        **inputs, 
        max_new_tokens=max_tokens,  # Máximo de tokens nuevos a generar
        use_cache=True,             # Usar caché para acelerar generación
        temperature=temperatura,     # Controla aleatoriedad (0.1-1.0)
        do_sample=True,             # Activar muestreo (vs greedy)
        pad_token_id=tokenizer.eos_token_id  # Token de padding
    )
    
    # Decodificar y extraer solo la respuesta
    respuesta_completa = tokenizer.batch_decode(outputs)[0]
    respuesta = respuesta_completa.split("### Respuesta:")[1].split("<|end_of_text|>")[0].strip()
    
    return respuesta

# Probar el modelo con una consulta nueva (no vista durante entrenamiento)
pregunta_test = "¿Qué fertilizante recomendarías para un lote de trigo en siembra directa en la zona de Venado Tuerto?"
respuesta = consultar_modelo(pregunta_test)

print("CONSULTA AGRÍCOLA:")
print(f"'{pregunta_test}'")
print("\nRESPUESTA DEL MODELO FINE-TUNEADO:")
print(f"'{respuesta}'")
print("\n" + "="*80)
print("NOTA: Esta respuesta fue generada por el modelo después del fine-tuning.")
print("Comparála con una respuesta del modelo base para ver la diferencia.")
````

### Paso 7: Guardar el Modelo

````python
# Guardar el modelo fine-tuneado para uso futuro

# Guardar los adaptadores LoRA (solo los parámetros nuevos)
# Esto es mucho más eficiente que guardar el modelo completo
model.save_pretrained("asistente_agricola_lora")  # Guarda solo las capas LoRA (~10-50MB)
tokenizer.save_pretrained("asistente_agricola_lora")  # Guarda el tokenizador

print("Modelo guardado exitosamente en 'asistente_agricola_lora/'")
print("\nArchivos guardados:")
print("- adapter_config.json: Configuración de LoRA")
print("- adapter_model.safetensors: Pesos de LoRA entrenados")
print("- tokenizer files: Tokenizador para procesar texto")

print("\n" + "="*60)
print("🎉 FINE-TUNING COMPLETADO EXITOSAMENTE! 🎉")
print("="*60)

print("\nPasos completados:")
print("✓ Cargamos modelo base Llama 3.1 8B")
print("✓ Configuramos LoRA para entrenamiento eficiente")
print("✓ Preparamos datos agrícolas argentinos")
print("✓ Entrenamos el modelo con nuestros ejemplos")
print("✓ Probamos el modelo con consultas nuevas")
print("✓ Guardamos el modelo para uso futuro")

print("\nPróximos pasos sugeridos:")
print("1. Expandir el dataset a 500-1000 ejemplos")
print("2. Evaluar sistemáticamente la calidad")
print("3. Optimizar hiperparámetros")
print("4. Crear interfaz web para consultas")
````

# 📖 Guía de Conceptos Clave

## Glosario Técnico

### **Fine-Tuning**
Proceso de entrenar un modelo preentrenado con datos específicos de un dominio para mejorar su rendimiento en tareas particulares. Es como darle educación especializada a alguien que ya tiene conocimientos generales.

### **LoRA (Low-Rank Adaptation)**
Técnica que permite fine-tuning eficiente modificando solo pequeñas "adaptaciones" en lugar de todos los parámetros del modelo. Reduce dramáticamente los recursos necesarios (memoria y tiempo).

### **Prompt Engineering**
Arte de diseñar instrucciones específicas para que un modelo genere las respuestas deseadas. No requiere entrenamiento adicional, pero debe repetirse en cada consulta.

### **RAG (Retrieval-Augmented Generation)**
Técnica que combina un modelo de lenguaje con una base de datos de conocimiento. El modelo busca información relevante antes de generar una respuesta.

### **Quantización 4-bit**
Técnica que reduce la precisión numérica de los parámetros del modelo de 16 bits a 4 bits, ahorrando aproximadamente 75% de memoria con mínima pérdida de calidad.

### **Learning Rate (Tasa de Aprendizaje)**
Parámetro que controla qué tan rápido aprende el modelo. Muy alto puede causar "olvido" del conocimiento previo; muy bajo puede ser demasiado lento.

### **Batch Size**
Número de ejemplos que el modelo procesa simultáneamente durante el entrenamiento. Afecta la estabilidad del entrenamiento y el uso de memoria.

### **Epochs**
Número de veces que el modelo pasa por todo el dataset de entrenamiento. Muy pocos epochs = aprendizaje insuficiente; demasiados = memorización (overfitting).

### **Overfitting**
Cuando el modelo memoriza los datos de entrenamiento en lugar de aprender patrones generalizables. Resulta en buen rendimiento en entrenamiento pero malo en datos nuevos.

### **Gradient Accumulation**
Técnica que permite simular batch sizes grandes acumulando gradientes de múltiples mini-batches antes de actualizar los parámetros. Útil cuando la memoria GPU es limitada.

## Conceptos Clave para Recordar

### **Calidad sobre Cantidad en Datos**
- 100 ejemplos excelentes > 1000 ejemplos mediocres
- Consistencia en formato y estilo
- Cobertura de la diversidad de casos reales

### **Cuándo Elegir Cada Técnica**
- **Prompt Engineering**: Tareas simples, resultados rápidos
- **RAG**: Información que cambia frecuentemente
- **Fine-Tuning**: Comportamiento específico y consistente

### **Evaluación del Modelo**
- **Cuantitativa**: Métricas automáticas (BLEU, ROUGE)
- **Cualitativa**: Revisión manual por expertos
- **Práctica**: Pruebas con usuarios reales

### **Eficiencia de LoRA**
- Solo 0.1-1% de parámetros se entrenan
- 60-80% menos uso de memoria
- Tiempos de entrenamiento 2-3x más rápidos
- Archivos de modelo muy pequeños (10-50MB vs 15GB+)

### Conceptos Fundamentales

**1. ¿Cuáles son las tres técnicas principales para personalizar un LLM y cuándo usar cada una?**

<details>
<summary>Respuesta</summary>

**Prompt Engineering**: Para tareas simples o puntuales que requieren resultados rápidos sin inversión en recursos. Ejemplo: análisis ocasional de datos climáticos.

**RAG**: Cuando necesitás información que se actualiza constantemente y tenés una buena base de conocimiento. Ejemplo: consultas sobre precios de commodities.

**Fine-Tuning**: Cuando necesitás comportamiento muy específico y consistente, tenés datos de calidad suficientes, y vas a usar el modelo repetidamente. Ejemplo: asistente agrónomo especializado.
</details>

<details>
<summary>Respuesta</summary>

- **Eficiencia de memoria**: 60-80% menos uso de memoria
- **Velocidad**: 2-3x más rápido
- **Archivos pequeños**: Modelos de 10-50MB vs 15GB+
- **Preservación**: Mantiene conocimiento base del modelo
- **Accesibilidad**: Permite fine-tuning en hardware modesto
</details>

**3. ¿Por qué es crucial la calidad sobre cantidad en los datos de entrenamiento?**

<details>
<summary>Respuesta</summary>

100 ejemplos excelentes son más valiosos que 1000 mediocres porque:
- El modelo aprende patrones más consistentes
- Mejor generalización a casos nuevos
- Menos riesgo de aprender comportamientos incorrectos
- Mayor eficiencia en tiempo de entrenamiento
- Resultados más predecibles y confiables
</details>

### Aspectos Técnicos

**4. Explicá qué hace cada uno de estos hiperparámetros:**
   - Learning Rate
   - Batch Size  
   - Epochs

<details>
<summary>Respuesta</summary>

**Learning Rate**: Controla qué tan rápido aprende el modelo. Muy alto = olvida conocimiento previo; muy bajo = aprende muy lento. Típico para LoRA: 2e-4.

**Batch Size**: Cuántos ejemplos procesa simultáneamente. Más grande = más estable pero requiere más memoria. Se debe maximizar según capacidad de GPU.

**Epochs**: Cuántas veces pasa por todos los datos. Muy pocos = aprendizaje insuficiente; demasiados = memorización (overfitting).
</details>

**5. ¿Qué es la quantización 4-bit y por qué es útil?**

<details>

La quantización 4-bit reduce la precisión numérica de los parámetros del modelo de 16 bits a 4 bits, logrando:
- 75% menos uso de memoria
- Posibilidad de ejecutar modelos grandes en hardware modesto
- Mínima pérdida de calidad en el rendimiento
- Democratización del acceso a modelos avanzados
</details>
